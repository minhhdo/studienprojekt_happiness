# Notebook for the Age Subgroup Analysis


```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# Read Data
data = pd.read_csv('../data/data_eqls_LOOSE_filtered.csv')

# Age is coded in five categories: 1–5
age_column = "Age"

df_age1 = data[data[age_column] == 1].copy()
df_age2 = data[data[age_column] == 2].copy()
df_age3 = data[data[age_column] == 3].copy()
df_age4 = data[data[age_column] == 4].copy()
df_age5 = data[data[age_column] == 5].copy()

print("Age category 1:", df_age1.shape)
print("Age category 2:", df_age2.shape)
print("Age category 3:", df_age3.shape)
print("Age category 4:", df_age4.shape)
print("Age category 5:", df_age5.shape)

# Labels for the five age categories
age_labels = {
    1: "Age 1 (18–24)",
    2: "Age 2 (25–34)",
    3: "Age 3 (35–49)",
    4: "Age 4 (50–64)",
    5: "Age 5 (65+)"
}
```

Mean Happiness level for all age categories


```python
data.groupby(age_column)["How happy are you?"].mean()
```

```python
data.groupby(age_column)["How happy are you?"].value_counts()
```

Happiness Distribution using Histogram plot


```python
for i, df in enumerate([df_age1, df_age2, df_age3, df_age4, df_age5], 1):
    plt.hist(df["How happy are you?"], alpha=0.5, label=age_labels[i], density=True)

plt.legend()
plt.title("Happiness Distribution by Age Categories")
plt.xlabel("Happiness")
plt.ylabel("Density")
plt.show()
```

```python
fig, axes = plt.subplots(5, 1, figsize=(6, 12), sharex=True)

dfs = [df_age1, df_age2, df_age3, df_age4, df_age5]

for i, (ax, df) in enumerate(zip(axes, dfs), 1):
    ax.hist(df["How happy are you?"], alpha=0.7, density=True)
    ax.set_title(age_labels[i])
    ax.set_ylabel("Density")
    ax.set_ylim(0, 0.4)

    mean = df["How happy are you?"].mean()
    ax.axvline(mean, linestyle='--', label=f"Mean = {mean:.2f}")
    ax.legend(loc="upper left")

axes[-1].set_xlabel("Happiness")

plt.tight_layout()
plt.show()
```

Happiness Distribution using Boxplot plot

```python
data.boxplot(column="How happy are you?", by=age_column)
plt.title("Happiness by Age Categories")
plt.suptitle("")
plt.xlabel("Age category")
plt.ylabel("Happiness")
plt.xticks(
    ticks=range(1, 6),
    labels=[age_labels[i] for i in range(1, 6)],
    rotation=30,
    ha="right"
)
plt.tight_layout()
plt.show()
```

## Model Training

```python
import importlib
from XGBoost import xgboost_machine_learning
importlib.reload(xgboost_machine_learning)
```

Model Settings

```python
my_target_column = "How happy are you?"
my_test_size = 0.2
my_random_state = 42

# Model-specific hyperparameters (baseline defaults)
my_n_estimators = 100
my_max_depth = None

# SHAP settings
my_number_of_shaply_rows = 5000
my_number_header_rows = 150
```

Run model training and shap

```python
age_groups = {
    "age1": df_age1,
    "age2": df_age2,
    "age3": df_age3,
    "age4": df_age4,
    "age5": df_age5
}
```

```python
for name, subgroup_data in age_groups.items():
    print(f"Running model for {name}...")

    my_target_column = 'How happy are you?'
    print("Shape before: ", subgroup_data.shape)

    # Drop 'How satisfied with...' columns
    # They have too much predictive value and there are similar questions
    # e.g. "Health condition", "Quality of education system?", "Feel left out of society?"
    columns_to_remove = [
        'How satisfied with present standard of living?',
        'How satisfied with family life?',
        'How satisfied with social life?',
        'How satisfied with health?',
        'How satisfied with economic situation in the country?',
        'How satisfied with accommodation?',
        'How satisfied with education?'
    ]

    subgroup_data = subgroup_data.drop(columns=columns_to_remove, errors='ignore').copy()
    print("Shape after: ", subgroup_data.shape)

    model_1B, shap_results_1B = xgboost_machine_learning.run_xgb_gridsearch_and_shap(
        df=subgroup_data,
        target_column=my_target_column,
        test_size=my_test_size,
        random_state=my_random_state,
        cv_folds=3,
        shap_rows=my_number_of_shaply_rows,
        number_header_rows=my_number_header_rows
    )

    shap_results_1B.to_csv(f"Results/age_{name}_shap_results_1B_optimized.csv", index=False)
```

### Plot Results

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import textwrap
```

```python
age_labels = {
    1: "Age 1 (18–24)",
    2: "Age 2 (25–34)",
    3: "Age 3 (35–40)",
    4: "Age 4 (50–64)",
    5: "Age 5 (65+)"
}

# Load Data
df_age1 = pd.read_csv("Results/age_age1_shap_results_1B_optimized.csv")
df_age2 = pd.read_csv("Results/age_age2_shap_results_1B_optimized.csv")
df_age3 = pd.read_csv("Results/age_age3_shap_results_1B_optimized.csv")
df_age4 = pd.read_csv("Results/age_age4_shap_results_1B_optimized.csv")
df_age5 = pd.read_csv("Results/age_age5_shap_results_1B_optimized.csv")

age_shap_dfs = [df_age1, df_age2, df_age3, df_age4, df_age5]

def plot_shap_large(df, title, max_questions=None):

    df_sorted = df.sort_values(by="Value", ascending=True).reset_index(drop=True)
    if max_questions:
        df_sorted = df_sorted.tail(max_questions).reset_index(drop=True)

    df_sorted["Questions"] = df_sorted["Questions"].apply(lambda x: "\n".join(textwrap.wrap(x, width=25)))

    n = len(df_sorted)
    plt.figure(figsize=(7, 8))
    bars = plt.barh(df_sorted["Questions"], df_sorted["Value"])

    plt.xlabel("Mean absolute SHAP value", color='black', fontsize=10)
    plt.title(title, color='black', fontsize=10)

    xlim_max = max(df["Value"].max() for df in age_shap_dfs)
    plt.xlim(0, xlim_max + 0.05)

    for bar in bars:
        width = bar.get_width()
        plt.text(width + 0.001,
                 bar.get_y() + bar.get_height() / 2,
                 f"{width:.3f}",
                 va='center', fontsize=10)

    plt.ylim(-0.5, n - 0.5)
    plt.tight_layout()
    plt.show()

# Show plots
plot_shap_large(df_age1, f"SHAP Values of Subgroup {age_labels[1]}", max_questions=15)
plot_shap_large(df_age2, f"SHAP Values of Subgroup {age_labels[2]}", max_questions=15)
plot_shap_large(df_age3, f"SHAP Values of Subgroup {age_labels[3]}", max_questions=15)
plot_shap_large(df_age4, f"SHAP Values of Subgroup {age_labels[4]}", max_questions=15)
plot_shap_large(df_age5, f"SHAP Values of Subgroup {age_labels[5]}", max_questions=15)
plt.show()
```
